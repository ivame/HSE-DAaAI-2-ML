Выразим фактическое значение $y_t$ через ошибку:

1. Доказательство для уровня ($\hat l_t$)

Начнем с рекуррентной формы:
$$ \hat l_t = \alpha (y_t - \hat s_{t-p}) + (1-\alpha) (\hat l_{t-1} + \hat b_{t-1}) $$
Раскроем скобки во втором слагаемом:
$$ \hat l_t = \alpha (y_t - \hat s_{t-p}) + \hat l_{t-1} + \hat b_{t-1} - \alpha (\hat l_{t-1} + \hat b_{t-1}) $$
Сгруппируем слагаемые с множителем $\alpha$:
$$ \hat l_t = \hat l_{t-1} + \hat b_{t-1} + \alpha \Big( y_t - \hat s_{t-p} - (\hat l_{t-1} + \hat b_{t-1}) \Big) $$
Перегруппируем выражение в скобках:
$$ \hat l_t = \hat l_{t-1} + \hat b_{t-1} + \alpha \Big( y_t - (\hat l_{t-1} + \hat b_{t-1} + \hat s_{t-p}) \Big) $$
Подставим определение ошибки $e_t$:
$$ \hat l_t = \hat l_{t-1} + \hat b_{t-1} + \alpha e_t $$
*Равенство доказано.*



2. Доказательство для тренда ($\hat b_t$)

Начнем с рекуррентной формы:
$$ \hat b_t = \beta (\hat l_{t} - \hat l_{t-1}) + (1-\beta) \hat b_{t-1} $$
Раскроем скобки во втором слагаемом:
$$ \hat b_t = \beta (\hat l_{t} - \hat l_{t-1}) + \hat b_{t-1} - \beta \hat b_{t-1} $$
Сгруппируем слагаемые с множителем $\beta$:
$$ \hat b_t = \hat b_{t-1} + \beta \Big( \hat l_{t} - \hat l_{t-1} - \hat b_{t-1} \Big) $$
Из доказанной выше формы корректировки для уровня ($\hat l_t = \hat l_{t-1} + \hat b_{t-1} + \alpha e_t$) следует, что:
$$ \hat l_{t} - \hat l_{t-1} - \hat b_{t-1} = \alpha e_t $$
Подставим это выражение в уравнение тренда:
$$ \hat b_t = \hat b_{t-1} + \beta (\alpha e_t) $$
$$ \hat b_t = \hat b_{t-1} + \alpha\beta e_t $$



3. Доказательство для сезонности ($\hat s_t$)

Начнем с рекуррентной формы:
$$ \hat s_t = \gamma (y_t - \hat l_{t}) + (1-\gamma) \hat s_{t-p} $$
Раскроем скобки во втором слагаемом:
$$ \hat s_t = \gamma (y_t - \hat l_{t}) + \hat s_{t-p} - \gamma \hat s_{t-p} $$
Сгруппируем слагаемые с множителем $\gamma$:
$$ \hat s_t = \hat s_{t-p} + \gamma \Big( y_t - \hat l_{t} - \hat s_{t-p} \Big) $$
Подставим выражение для $y_t$ из формулы $(*)$ в скобки:
$$ y_t - \hat l_{t} - \hat s_{t-p} = (\hat l_{t-1} + \hat b_{t-1} + \hat s_{t-p} + e_t) - \hat l_{t} - \hat s_{t-p} $$
Сократим $\hat s_{t-p}$ и перегруппируем:
$$ = \hat l_{t-1} + \hat b_{t-1} + e_t - \hat l_{t} $$
$$ = e_t - (\hat l_{t} - \hat l_{t-1} - \hat b_{t-1}) $$
Как показано в пункте 2, выражение в скобках равно $\alpha e_t$. Подставим это:
$$ = e_t - \alpha e_t = (1 - \alpha)e_t $$
Вернемся к уравнению сезонности и подставим полученный результат:
$$ \hat s_t = \hat s_{t-p} + \gamma \Big( (1 - \alpha)e_t \Big) $$
$$ \hat s_t = \hat s_{t-p} + \gamma(1-\alpha)e_t $$>

Use MAPE as a loss function.

```python
import math
import numpy as np
import pandas as pd
import utils
import re

def WintersExponentialSmoothing(x, h, Params):
    x_vals = np.array(x)
    T = len(x_vals)

    alpha = Params['alpha']
    gamma = Params['gamma']
    p = int(Params['seasonality_period'])

    FORECAST = np.full(T + h, np.nan)
    l = np.nan
    s = np.full(p, np.nan)

    for cntr in range(T):
        if not math.isnan(x_vals[cntr]):
            if math.isnan(l):
                l = x_vals[cntr]

            if math.isnan(s[cntr % p]):
                s[cntr % p] = 0.0

            l_new = alpha * (x_vals[cntr] - s[cntr % p]) + (1 - alpha) * l
            s[cntr % p] = gamma * (x_vals[cntr] - l_new) + (1 - gamma) * s[cntr % p]
            l = l_new

        if not math.isnan(l):
            FORECAST[cntr + h] = l + s[(cntr + h) % p]

    return FORECAST

utils.WintersExponentialSmoothing = WintersExponentialSmoothing
```

```python
import math
import numpy as np

def TheilWageExponentialSmoothing(x, h, Params):
    x_vals = np.array(x)
    T = len(x_vals)
    alpha = Params['alpha']
    beta = Params['beta']
    gamma = Params['gamma']
    p = int(Params['seasonality_period'])

    FORECAST = [np.nan] * (T + h)

    l = np.nan
    b = np.nan
    s = []

    for i in range(p):
        if i < T:
            s.append(x_vals[i])
        else:
            s.append(0.0)

    for cntr in range(T):
        if not math.isnan(x_vals[cntr]):
            if math.isnan(l):
                l = x_vals[cntr]
            if math.isnan(b):
                b = 0.0

            if cntr < p:
                l_old = l
                l = alpha * (x_vals[cntr] - s[cntr]) + (1 - alpha) * (l + b)
                b = beta * (l - l_old) + (1 - beta) * b
            else:
                l_old = l
                s.append(gamma * (x_vals[cntr] - l) + (1 - gamma) * s[cntr - p])
                l = alpha * (x_vals[cntr] - s[cntr - p]) + (1 - alpha) * (l + b)
                b = beta * (l - l_old) + (1 - beta) * b

        if not math.isnan(l) and len(s) > 0:
            idx = cntr + h - (1 + h // p) * p
            if -len(s) <= idx < len(s):
                FORECAST[cntr + h] = l + b + s[idx]
            else:
                FORECAST[cntr + h] = l + b + s[(cntr + h) % p]

    return FORECAST

utils.TheilWageExponentialSmoothing = TheilWageExponentialSmoothing
```

```python
import math
import re
import numpy as np
import pandas as pd

import math
import numpy as np

def MultiplicativeSeasonalityES(x, h, Params):
    x_vals = np.array(x, dtype=float)
    T = len(x_vals)
    alpha = Params['alpha']
    beta = Params['beta']
    gamma = Params['gamma']
    p = int(Params['seasonality_period'])

    FORECAST = [np.nan] * (T + h)
    l = np.nan
    b = np.nan
    s = [1.0] * p

    for cntr in range(T):
        if not math.isnan(x_vals[cntr]):
            if math.isnan(l):
                l = x_vals[cntr] if x_vals[cntr] != 0 else 1.0
                b = 0.0
                for i in range(p):
                    if i < T and not math.isnan(x_vals[i]):
                        s[i] = x_vals[i] / l if l != 0 else 1.0
                    else:
                        s[i] = 1.0

            s_old = s[cntr % p]
            if s_old == 0 or math.isnan(s_old):
                s_old = 1.0

            l_new = alpha * (x_vals[cntr] / s_old) + (1 - alpha) * (l + b)

            if l_new == 0 or math.isnan(l_new):
                l_new = l if not math.isnan(l) else 1.0

            b = beta * (l_new - l) + (1 - beta) * b
            s[cntr % p] = gamma * (x_vals[cntr] / l_new) + (1 - gamma) * s_old
            l = l_new

        if not math.isnan(l):
            idx = (cntr + h) % p
            s_idx = s[idx] if not math.isnan(s[idx]) else 1.0
            FORECAST[cntr + h] = (l + h * b) * s_idx

    return FORECAST

utils.MultiplicativeSeasonalityES = MultiplicativeSeasonalityES
```

Если горизонт прогнозирования h равен периоду сезонности p, результаты не будут одинаковыми для аддитивной и мультипликативной моделей.

Причины:
1. Разная природа сезонности. Аддитивная модель предполагает, что амплитуда сезонных колебаний постоянна во времени: прогноз = уровень + тренд + сезонность. Мультипликативная модель предполагает, что амплитуда пропорциональна уровню ряда: прогноз = (уровень + тренд) × сезонность.

2. Масштабирование колебаний. Если ряд имеет растущий тренд, то при h=p мультипликативная модель даст сезонные всплески, которые растут вместе с трендом (например, если уровень удвоился, то и сезонный пик удвоится). Аддитивная модель даст те же абсолютные отклонения, что и в прошлом, независимо от текущего уровня.
